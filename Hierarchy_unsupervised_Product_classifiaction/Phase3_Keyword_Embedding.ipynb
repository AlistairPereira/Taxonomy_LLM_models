```python
import pandas as pd
import numpy as np
```

```python
df_all = pd.read_json("phase_2_keywords.json", lines=True)
df_all.head()
```

```python
X = np.array(df_all["embedding"].tolist(), dtype="float32")
print("Embedding matrix shape:", X.shape)
X
```

```python
from sklearn.preprocessing import normalize
X_norm = normalize(X)   # row-wise L2 norm
print("Normalized matrix shape:", X_norm.shape)
```

```python
X_norm
```

```python
np.save("phase3_embeddings.npy", X_norm)                     # for clustering
df_all.to_parquet("phase3_table.parquet", index=False)       # fast tabular I/O
```

```python
print("Phase 3 complete.")
print("Samples — len(embedding[0]):", len(df_all['embedding'].iloc[0]))
print("Columns available:", list(df_all.columns))
```
